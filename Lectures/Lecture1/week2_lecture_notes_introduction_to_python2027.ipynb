BOLUM 10: SARKAC SIMULASYONU / SECTION 10: SIMPLE PENDULUM MOTION SIMULATION

TR: L uzunlugundaki bir ipin ucundaki m kutlesinin hareketini Euler yontemi ve bir for dongusu kullanarak adim adim simule edecegiz.
EN: We will simulate the step-by-step motion of a mass suspended from a string of length L using Euler's method and a for loop.

```python
import math
import matplotlib.pyplot as plt

# --- Parameters ---
g = 9.81         # Gravity (m/s^2)
L = 1.0          # Length of the pendulum string (meters)
dt = 0.05        # Time step size (seconds)
total_time = 10  # Total simulation time (seconds)

# --- Initial Conditions ---
theta = math.radians(30) # Initial angle of 30 degrees converted to radians
omega = 0.0              # Initial angular velocity (starting from rest)

# Lists to record time and angle history for plotting
time_history = []
theta_history = []

current_time = 0.0
n_steps = int(total_time / dt)

# --- Simulation Loop using Euler's Method ---
for step in range(n_steps):
    # Record history
    time_history.append(current_time)
    theta_history.append(math.degrees(theta)) # Convert back to degrees for plotting

    # Calculate angular acceleration: alpha = - (g / L) * sin(theta)
    alpha = - (g / L) * math.sin(theta)

    # Update velocity: omega = omega + alpha * dt
    omega += alpha * dt

    # Update position: theta = theta + omega * dt
    theta += omega * dt

    # Update current time
    current_time += dt

# --- Plotting the Results ---
plt.figure(figsize=(10, 5))
plt.plot(time_history, theta_history, label="Angular Displacement ($\theta$)", color="royalblue", linewidth=2)
plt.title("Simple Pendulum Motion Simulation (Euler's Method)", fontsize=14)
plt.xlabel("Time (seconds)", fontsize=12)
plt.ylabel("Angle (degrees)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```

FARKLI YERCEKIMI IVMELERI / COMPARING DIFFERENT GRAVITY CONDITIONS

TR: Yercekimi ivmesinin (g) sarkaç hareketine etkisini görmek için Dünya, Mars ve Ay senaryolarini karsilastiriyoruz.
EN: To observe the effect of gravity (g) on pendulum motion, we compare Earth, Mars, and Moon environments.

```python
import math
import matplotlib.pyplot as plt

# --- Shared Parameters ---
L = 1.0          # Length of the pendulum string (meters)
dt = 0.05        # Time step size (seconds)
total_time = 15  # Increased simulation time to 15 seconds to see trends clearly

# --- Gravitational Scenarios ---
gravities = {
    "Earth (9.81 m/s²)": 9.81,
    "Mars (3.71 m/s²)": 3.71,
    "Moon (1.62 m/s²)": 1.62
}

plt.figure(figsize=(12, 6))

# Loop through each gravity condition and simulate
for label, g_val in gravities.items():
    # Reset initial conditions
    theta = math.radians(30) # 30 degrees initial position
    omega = 0.0              # starting from rest

    time_history = []
    theta_history = []
    current_time = 0.0
    n_steps = int(total_time / dt)

    # Run simulation
    for step in range(n_steps):
        time_history.append(current_time)
        theta_history.append(math.degrees(theta))

        alpha = - (g_val / L) * math.sin(theta)
        omega += alpha * dt
        theta += omega * dt
        current_time += dt

    # Plot this specific scenario
    plt.plot(time_history, theta_history, label=label, linewidth=2)

# --- Plot styling ---
plt.title("Comparison of Pendulum Motion Under Different Gravity Conditions", fontsize=14)
plt.xlabel("Time (seconds)", fontsize=12)
plt.ylabel("Angle (degrees)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(title="Environments")
plt.show()
```

KUCUK ACI YAKLASIMI / THE SMALL-ANGLE APPROXIMATION

TR: Salinim acisi cok kucukken sin(theta) yaklasik esittir theta varsayimi yapilabilir. Bu kodla olusan hata payini yuzdesel olarak inceleyecegiz.
EN: When the oscillation angle is very small, we can assume sin(theta) is approximately equal to theta. We will analyze the percentage error of this approximation.

```python
import math
import pandas as pd

# Let's test different angles in degrees
test_angles_deg = [1, 5, 10, 15, 30, 45, 90]
data_comparison = []

for deg in test_angles_deg:
    theta_rad = math.radians(deg)
    sin_theta = math.sin(theta_rad)

    # Percentage error formula: |(Approx - Actual) / Actual| * 100
    pct_error = abs(theta_rad - sin_theta) / sin_theta * 100

    data_comparison.append({
        "Angle (Degrees)": f"{deg}°",
        "Angle (Radians) [Approx]": round(theta_rad, 5),
        "sin(Angle) [Actual]": round(sin_theta, 5),
        "Error (%)": f"{pct_error:.3f}%"
    })

# Convert to a DataFrame to display as a structured table
df_approx = pd.DataFrame(data_comparison)
display(df_approx)
```

BASIT HARMONIK HAREKET / WHY IS IT CALLED 'SIMPLE' HARMONIC MOTION?

TR: Geri cagirici kuvvetin dogrusal olmasi (F = -kx), tek bir sabit frekans bulunmasi ve enerji kaybinin ihmal edilmesi nedeniyle 'Basit' denir.
EN: It is called 'Simple' because the restoring force is linear (F = -kx), there is a single constant frequency, and energy loss is neglected.

```python
import math
import matplotlib.pyplot as plt

# --- Parameters ---
g = 9.81
L = 1.0
dt = 0.02
total_time = 6.0

# Start at a large angle (90 degrees) to highlight the difference
theta_init = math.radians(90)

# --- 1. Ideal Simple Harmonic Motion (Linearized: sin(theta) ~ theta) ---
# Under SHM, the analytical solution is theta(t) = theta_init * cos(sqrt(g/L) * t)
time_shm = []
theta_shm = []
omega_shm = math.sqrt(g / L)

# --- 2. Actual Non-Linear Pendulum (Using Euler's Method with sin(theta)) ---
time_actual = []
theta_actual = []
theta_curr = theta_init
omega_curr = 0.0
t_curr = 0.0

n_steps = int(total_time / dt)
for step in range(n_steps):
    # Record SHM solution
    time_shm.append(t_curr)
    theta_shm.append(math.degrees(theta_init * math.cos(omega_shm * t_curr)))

    # Record Actual solution
    time_actual.append(t_curr)
    theta_actual.append(math.degrees(theta_curr))

    # Physics step for actual motion
    alpha = - (g / L) * math.sin(theta_curr)
    omega_curr += alpha * dt
    theta_curr += omega_curr * dt

    t_curr += dt

# --- Plotting the Comparison ---
plt.figure(figsize=(12, 6))
plt.plot(time_shm, theta_shm, '--', label="Simple Harmonic Motion (Approximate)", color="red", linewidth=2)
plt.plot(time_actual, theta_actual, label="Actual Large-Angle Motion (Real Pendulum)", color="darkblue", linewidth=2)
plt.title("Simple Harmonic Motion vs. Real Large-Angle Pendulum (90°)", fontsize=14)
plt.xlabel("Time (seconds)", fontsize=12)
plt.ylabel("Angle (degrees)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```

SIMULASYONLARI FONKSIYON YAPISINA DONUSTURME / MODULARIZING WITH FUNCTIONS

TR: Kod tekrarini onlemek icin sarkaç simülasyonunu 'simulate_pendulum' isimli tek bir fonksiyonda topluyoruz.
EN: To prevent code duplication, we encapsulate the simulation steps into a single reusable function called 'simulate_pendulum'.

```python
import math
import matplotlib.pyplot as plt

def simulate_pendulum(g=9.81, L=1.0, theta_init_deg=30.0, total_time=15.0, dt=0.05):
    """
    Simulates a simple pendulum using Euler's method.

    Parameters:
      g: Gravitational acceleration (m/s^2)
      L: Length of the string (meters)
      theta_init_deg: Starting angle in degrees
      total_time: Simulation duration (seconds)
      dt: Time step (seconds)

    Returns:
      time_history, theta_history (in degrees)
    """
    theta = math.radians(theta_init_deg)
    omega = 0.0
    current_time = 0.0
    n_steps = int(total_time / dt)

    time_history = []
    theta_history = []

    for step in range(n_steps):
        time_history.append(current_time)
        theta_history.append(math.degrees(theta))

        alpha = - (g / L) * math.sin(theta)
        omega += alpha * dt
        theta += omega * dt
        current_time += dt

    return time_history, theta_history
```

BOLUM 11: FONKSIYON TANIMLAMA / SECTION 11: DEFINING CUSTOM FUNCTIONS

TR: Kendi fonksiyonlarinizi 'def' anahtar kelimesiyle tanimlayarak tekrar tekrar kullanabilirsiniz.
EN: You can define your own reusable code blocks using Python's 'def' keyword.

```python
# Örnek 1: Sayının karesini alan temel bir fonksiyon
def karesini_al(sayi):
    """Girdi olarak alınan sayının karesini geri döndürür."""
    return sayi ** 2

sonuc = karesini_al(6)
print("1) 6 sayısının karesi:", sonuc)

# Örnek 2: Dikdörtgenin alanını hesaplayan fonksiyon (Çok parametreli)
def alan_hesapla(kisa_kenar, uzun_kenar):
    """Kısa ve uzun kenarı çarparak dikdörtgen alanını bulur."""
    return kisa_kenar * uzun_kenar

dikdortgen_alani = alan_hesapla(5, 10)
print("2) Kenarları 5 ve 10 olan dikdörtgenin alanı:", dikdortgen_alani)

# Örnek 3: Sıcaklığı Celsius'tan Fahrenheit'a çeviren fonksiyon
def celsius_to_fahrenheit(celsius):
    """Celsius dereceyi Fahrenheit dereceye dönüştürür."""
    return (celsius * 9/5) + 32

f_sicaklik = celsius_to_fahrenheit(25)
print("3) 25°C sıcaklığın Fahrenheit karşılığı:", f_sicaklik)

# Örnek 4: Sayının tek mi çift mi olduğunu kontrol eden fonksiyon (Koşul ve Boolean)
def cift_mi(sayi):
    """Sayı çift ise True, tek ise False döner."""
    if sayi % 2 == 0:
        return True
    else:
        return False

print("4) 14 sayısı çift mi?:", cift_mi(14))
print("   15 sayısı çift mi?:", cift_mi(15))

# Örnek 5: İsim alıp kişiye özel karşılama mesajı üreten fonksiyon
def karsilama_mesaji(isim):
    """Girilen isme özel bir selamlama metni hazırlar."""
    return f"Merhaba {isim}, Python dünyasına hoş geldin! 👋"

mesaj = karsilama_mesaji("Ahmet")
print("5)", mesaj)
```

KULLANICIDAN INTERAKTIF GIRDI ALMA / INTERACTIVE USER INPUT

TR: input() fonksiyonu ile kullanicidan bilgi alabilir ve bunu sayisal tiplere donusturerek fonksiyonlarimizda kullanabiliriz.
EN: We can read interactive user inputs using the input() function and convert them to numeric types for our functions.

```python
# 1. Ask the user for an integer input
user_input = input("Lütfen bir tam sayı giriniz (Enter an integer): ")

# 2. Convert string input to integer
try:
    selected_number = int(user_input)

    # 3. Call our previously defined functions
    square_result = karesini_al(selected_number)
    even_odd_status = cift_mi(selected_number)

    # 4. Print structured output
    print("\n--- Sonuçlar (Results) ---")
    print(f"Girdiğiniz sayı (Your number): {selected_number}")
    print(f"Sayının karesi (Square): {square_result}")

    if even_odd_status:
        print("Bu sayı ÇİFT (Even) bir sayıdır.")
    else:
        print("Bu sayı TEK (Odd) bir sayıdır.")

except ValueError:
    print("Hata: Lütfen geçerli bir tam sayı girdiğinizden emin olun! (Error: Please enter a valid integer!)")
```

TR: Simdi yeni yazdigimiz 'simulate_pendulum' fonksiyonunu kullanarak tum deneyleri tek seferde calistirip karsilastiralim.
EN: Now we will run all scenarios at once using our new 'simulate_pendulum' function and compare them on a single clean layout.

```python
# Creating subplots to show all 3 studies side-by-side cleanly
fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

# --- Case 1: Different Gravities ---
axes[0].set_title("1. Varying Gravity (g)")
gravities = {"Earth (9.81)": 9.81, "Mars (3.71)": 3.71, "Moon (1.62)": 1.62}
for name, g_val in gravities.items():
    t, th = simulate_pendulum(g=g_val, L=1.0, theta_init_deg=30.0)
    axes[0].plot(t, th, label=name)
axes[0].set_xlabel("Time (s)")
axes[0].set_ylabel("Angle (degrees)")
axes[0].legend()
axes[0].grid(True, linestyle=":")

# --- Case 2: Different Lengths ---
axes[1].set_title("2. Varying Length (L)")
lengths = {"Short (0.5m)": 0.5, "Medium (1.0m)": 1.0, "Long (2.0m)": 2.0}
for name, L_val in lengths.items():
    t, th = simulate_pendulum(g=9.81, L=L_val, theta_init_deg=30.0)
    axes[1].plot(t, th, label=name)
axes[1].set_xlabel("Time (s)")
axes[1].legend()
axes[1].grid(True, linestyle=":")

# --- Case 3: Different Initial Angles ---
axes[2].set_title("3. Varying Initial Angle (θ₀)")
angles = {"Small (10°)": 10.0, "Medium (45°)": 45.0, "Large (90°)": 90.0}
for name, angle_val in angles.items():
    t, th = simulate_pendulum(g=9.81, L=1.0, theta_init_deg=angle_val)
    axes[2].plot(t, th, label=name)
axes[2].set_xlabel("Time (s)")
axes[2].legend()
axes[2].grid(True, linestyle=":")

plt.tight_layout()
plt.show()
```
